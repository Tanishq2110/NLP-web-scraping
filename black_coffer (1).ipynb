```python
import os
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Function to extract article content
def extract_blackcoffer_article(url):
    try:
        # Fetch the webpage
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Extract the article title
            title_tag = soup.find('h1', class_='entry-title')  # Adjust class if necessary
            title = title_tag.get_text(strip=True) if title_tag else "No title found"
            
            # Extract the article text
            content_div = soup.find('div', class_='td-post-content')  # Adjust class if necessary
            if content_div:
                # Extract only paragraphs
                paragraphs = content_div.find_all('p')
                article_text = ' '.join(p.get_text(strip=True) for p in paragraphs)
            else:
                article_text = "No article content found"
            
            return title, article_text
        else:
            print(f"Failed to fetch URL: {url}, Status Code: {response.status_code}")
            return None, None
    except Exception as e:
        print(f"Error fetching URL {url}: {e}")
        return None, None

# Read the input file
input_file = r"C:\Users\tanis\Downloads\Input.xlsx"  # Replace with your input file path
output_dir = 'extracted_articles'
os.makedirs(output_dir, exist_ok=True)  # Create directory for saving articles

data = pd.read_excel(input_file, engine='openpyxl')

# Process each URL
for _, row in data.iterrows():
    url_id = row['URL_ID']
    url = row['URL']
    
    print(f"Processing URL_ID: {url_id}, URL: {url}")
    title, article_text = extract_blackcoffer_article(url)
    
    if title and article_text:
        # Save the article to a text file
        file_name = os.path.join(output_dir, f"{url_id}.txt")
        with open(file_name, 'w', encoding='utf-8') as file:
            file.write(f"Title: {title}\n\n{article_text}")
        print(f"Saved article to {file_name}")
    else:
        print(f"Failed to extract content for URL_ID: {url_id}")

print("Processing complete!")
```

```python
# Directory containing the text files
text_files_dir = r"C:\Users\tanis\extracted_articles"  # Replace with the directory containing your text files

# Keyword to identify and remove content after
keyword = "Summarized:"

# Iterate over all text files in the directory
for file_name in os.listdir(text_files_dir):
    if file_name.endswith(".txt"):
        file_path = os.path.join(text_files_dir, file_name)
        
        # Read the file content
        with open(file_path, "r", encoding="utf-8") as file:
            lines = file.readlines()
        
        # Identify starting point (last 5 lines)
        start_index = max(0, len(lines) - 5)
        last_5_lines = lines[start_index:]
        
        # Check if the keyword exists in the last 5 lines
        cleaned_lines = []
        keyword_found = False
        for line in last_5_lines:
            if keyword_found:
                continue  # Skip lines after the keyword is found
            if keyword in line:
                keyword_found = True
                line = line.split(keyword)[0]  # Keep text before the keyword
            cleaned_lines.append(line)
        
        # Combine the earlier lines with the cleaned last 5 lines
        final_lines = lines[:start_index] + cleaned_lines
        
        # Write the cleaned content back to the file
        with open(file_path, "w", encoding="utf-8") as file:
            file.writelines(final_lines)

print("Content after 'Summarized:' in the last 5 lines has been removed from all files.")
```

```python
import os
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
import string

# Download NLTK data
nltk.download('punkt')


# File paths
output_file_path = r"C:\Users\tanis\Downloads\Output Data Structure.xlsx"  # Existing output file with the specified structure
cleaned_files_directory = r"C:\Users\tanis\cleaned_articles"  # Directory with cleaned files
positive_words_file = r"C:\Users\tanis\Downloads\MasterDictionary-20241211T122151Z-001\MasterDictionary\positive-words.txt"  # Your positive words list
negative_words_file = r"C:\Users\tanis\Downloads\MasterDictionary-20241211T122151Z-001\MasterDictionary\negative-words.txt"  # Your negative words list

# Load positive and negative words
def load_words(file_path):
    with open(file_path, "r", encoding="utf-8", errors="ignore") as file:
        return set(word.strip().lower() for word in file if word.strip() and not word.startswith(";"))

positive_words = load_words(positive_words_file)
negative_words = load_words(negative_words_file)

# Load the output DataFrame
output_df = pd.read_excel(output_file_path)

# Functions to calculate scores
def calculate_positive_score(text, positive_words):
    tokens = word_tokenize(text.lower())
    return sum(1 for word in tokens if word in positive_words)

def calculate_negative_score(text, negative_words):
    tokens = word_tokenize(text.lower())
    return sum(1 for word in tokens if word in negative_words)

def calculate_polarity_score(positive_score, negative_score):
    return (positive_score - negative_score) / ((positive_score + negative_score) + 0.000001)

def calculate_subjectivity_score(positive_score, negative_score, text):
    total_words = len(word_tokenize(text))
    return (positive_score + negative_score) / (total_words + 0.000001)

# Process each article
for index, row in output_df.iterrows():
    article_id = row['URL_ID']  # Correct column name
    filepath = os.path.join(cleaned_files_directory, f"{article_id}.txt")
    try:
        with open(filepath, 'r', encoding='utf-8', errors="ignore") as file:
            text = file.read()
            
            # Perform text analysis
            positive_score = calculate_positive_score(text, positive_words)
            negative_score = calculate_negative_score(text, negative_words)
            polarity_score = calculate_polarity_score(positive_score, negative_score)
            subjectivity_score = calculate_subjectivity_score(positive_score, negative_score, text)
            
            # Update the DataFrame
            output_df.loc[index, 'POSITIVE SCORE'] = positive_score
            output_df.loc[index, 'NEGATIVE SCORE'] = negative_score
            output_df.loc[index, 'POLARITY SCORE'] = polarity_score
            output_df.loc[index, 'SUBJECTIVITY SCORE'] = subjectivity_score
    except FileNotFoundError:
        print(f"File not found for URL_ID: {article_id}")

# Save the updated DataFrame
output_df.to_excel(output_file_path, index=False)
print("Updated output file saved.")
```

```python
# Ensure NLTK sentence tokenizer is downloaded
nltk.download('punkt')

# Load the output DataFrame
output_file_path = r"C:\Users\tanis\Downloads\Output Data Structure.xlsx"  # Existing output file with the specified structure
cleaned_files_directory = r"C:\Users\tanis\cleaned_articles"  # Directory with cleaned files


# Load the output file
output_df = pd.read_excel(output_file_path)

# Function to analyze text and compute metrics
def analyze_text_metrics(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        text = file.read()

    # Split into sentences
    sentences = nltk.sent_tokenize(text)
    total_sentences = len(sentences)

    # Split into words
    words = text.split()
    total_words = len(words)

    # Count syllables
    total_syllables = sum(syllables.estimate(word) for word in words)

    # Count total characters (excluding spaces)
    total_characters = sum(len(word) for word in words)

    # Count complex words
    complex_words = sum(1 for word in words if syllables.estimate(word) > 2)

    # Compute metrics
    avg_sentence_length = total_words / total_sentences if total_sentences > 0 else 0
    percentage_complex_words = (complex_words / total_words) if total_words > 0 else 0
    fog_index = 0.4 * (avg_sentence_length + percentage_complex_words)
    syllables_per_word = total_syllables / total_words if total_words > 0 else 0
    avg_word_length = total_characters / total_words if total_words > 0 else 0
    word_count = total_words
    

    return {
        "avg_sentence_length": avg_sentence_length,
        "percentage_complex_words": percentage_complex_words * 100,  # Convert to percentage
        "fog_index": fog_index,
        "syllables_per_word": syllables_per_word,
        "avg_word_length": avg_word_length,
        "word_count": word_count,
    }



# Process and update the output file
for index, row in output_df.iterrows():
    url_id = row['URL_ID']
    file_name = f"{url_id}.txt"
    file_path = os.path.join(cleaned_files_directory, file_name)

    # Check if file exists
    if os.path.exists(file_path):
        metrics = analyze_text_metrics(file_path)

        # Update the dataframe with calculated metrics
        output_df.at[index, 'AVG SENTENCE LENGTH'] = metrics['avg_sentence_length']
        output_df.at[index, 'PERCENTAGE OF COMPLEX WORDS'] = metrics['percentage_complex_words']
        output_df.at[index, 'FOG INDEX'] = metrics['fog_index']
        output_df.at[index, 'SYLLABLE PER WORD'] = metrics['syllables_per_word']
        output_df.at[index, 'AVG WORD LENGTH'] = metrics['avg_word_length']
        output_df.at[index, 'WORD COUNT'] = metrics['word_count']
    else:
        print(f"File {file_name} not found. Skipping.")

# Save the updated dataframe back to the output file
output_df.to_excel(output_file_path, index=False)
print(f"Updated metrics saved to {output_file_path}")
```
